```python
#Types of Functions

#1. Function with Return
def square(n):
    return n * n

print(square(5))

#2. Function without Return
def greet(name):
    print("Hello", name)

greet("Amit")

#3. Multiple Parameters
def multiply(a, b):
    return a * b

print(multiply(3, 4))

#4. Default Parameters
def greet(name="User"):
    return f"Hello {name}"

print(greet())
print(greet("Rahul"))

#5. Function + List
def get_even(nums):
    return [n for n in nums if n % 2 == 0]

print(get_even([10, 15, 20, 25]))
```

```python
#Q1: Create function to return square of number
def square(n):
    return n*n
print(square(5))

#Q2: Function to return sum of list
def sum_of_list(nums):
    return sum(nums)
print(sum_of_list([1,2,3,4,5]))

#Q3: Function to return max number from list (without using max())
def max_num(nums):
    m=nums[0]
    for i in nums:
        if i>m:
            m=i
    return m

print(max_num([5,4,9,8]))

#Q4: Function to return only even numbers from list
def even(nums):
    return [num for num in nums if num %2==0]
print(even([1,2,3,4,5,6,7,8,9]))

#Q5: Function: --input → list ----output → ["Even", "Odd", ...]
def oddeven(nums):
    return ["Even" if num % 2 ==0 else "Odd" for num in nums]
print(oddeven([1,2,3,4,5,6]))

#Q6: # Function:----input → list ,---- output → square only even numbers, keep odd same
def ls(nums):
    return [num**2 if num % 2==0 else num for num in nums]

print(ls([1,2,3,4,5,6,7]))


```

```python
#Q7 Function: input → list -----output → count of even numbers
def count_even(nums):
    count=0
    for i in nums:
        if i % 2==0:
            count+=1
    return count
print(count_even([14,15,12,1,4,9,6]))

#Q8: Function: input → list ----output → sum of only odd numbers
def sumodd(nums):
    total=0
    for i in nums:
        if i%2 != 0:
            total+=i
    return total
print(sumodd([14,15,12,1,4,9,6]))

#Q9:# Function: input → list ----output → second largest number
def secondlar(nums):
    fmax= float(-inf)
    smax= float(-inf)
    for i in nums:
        if i>fmax:
            smax=fmax
            fmax=i
    return smax

print(secondlar([5,3,7,1]))
```
