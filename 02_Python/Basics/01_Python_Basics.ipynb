# TASK SET (Day 1 Practice)

```python
#Q1: Even Numbers: Print only even numbers
nums = [10, 15, 20, 25, 30, 35]
for i in nums:
    if i%2==0:
        print(i)
```

```python
#Q2: Sum of List
total=0
for i in nums:
    total+=i
print("total sum",total)
```

```python
#Q3:Find max number manually
maximum=nums[0]
for i in nums:
    if i>maximum:
        maximum=i
print("max number",maximum)
```

```python
#Q4 — Dictionary Practice
emp = {
    "name": "Rahul",
    "salary": 60000,
    "city": "Delhi"
}
print(emp.keys())
print(emp.values())
for key, value in emp.items():
    print(key, value)
```

```python
# Function is even to check even or not
def is_even(n):
    return n % 2 == 0
print(is_even(7))
```
