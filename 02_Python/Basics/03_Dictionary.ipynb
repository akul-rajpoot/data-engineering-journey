```python
#🔥 2. Basic Example
emp = {
    "name": "Amit",
    "salary": 50000,
    "city": "Delhi"
}

print(emp["name"])   # Amit
```

```python
#🔥 3. Loop Through Dictionary
for key, value in emp.items():
    print(key, value)
```

```python
#🔥 4. Add / Update Values
emp["age"] = 25          # add
emp["salary"] = 60000   # update
```

```python
#🔥 5. Check Key Exists
if "name" in emp:
    print("Exists")
```

```python
#Q1:
k = ["apple", "banana", "apple", "orange", "banana"]
m = {}
for i in k:
    m[i] = m.get(i, 0) + 1
print(m)
```

```python
#Q1: Count frequency of numbers
nums = [1,2,2,3,3,3,4]
result={}
for i in nums:
    result[i]=result.get(i,0)+1
print(result)
```

```python
#Q2: Find character frequency
text = "aabbccc"
result={}
for i in text:
    result[i]=result.get(i,0)+1
print(result)
```

```python
#Q:3 Group numbers into even and odd
# Output:
# {"even": [...], "odd": [...]}
m=[1,2,3,4,5,6,7,8]
result={"Even":[],"Odd":[]}
for i in m:
    if i %2==0:
        result["Even"].append(i)
    else:
        result["Odd"].append(i)
print(result)
    
```

```python
#Q4: Find most frequent element in list
ls=[1,2,3,4,2,3,2,2,3,4]
result={}
count,num=0,''
for i in ls:
    result[i]=result.get(i,0)+1
    if count<result[i]:
        count=result[i]
        num=i
print(result)
print(num,count)
```

```python
# Input:

data = [
    {"name": "A", "dept": "IT"},
    {"name": "B", "dept": "HR"},
    {"name": "C", "dept": "IT"}
]
# Output: {"IT": ["A", "C"], "HR": ["B"]}
result={}
for i in data:
    result.setdefault(i["dept"], []).append(i["name"])
print(result)
    
    
```
