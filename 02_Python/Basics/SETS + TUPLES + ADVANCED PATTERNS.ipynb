```python
#Set:Unordered collection of UNIQUE elements
#✔ Remove duplicates
#✔ Fast lookup (O(1))
#✔ Used in filtering / joins
nums = [1,2,2,3,3,4]
s = set(nums)
print(s)
```

```python
#Set Operations

#1:Union
a = {1,2,3}
b = {3,4,5}
print(a | b)

#2: Intersection
print(a & b)

#3: Difference
print(a - b)

```

```python
#First Non-Repeating Element
data = ["a", "b", "a", "c", "b"]
freq={}
for i in data:
    freq[i]=freq.get(i,0)+1
for i in freq:
    if freq[i]==1:
        print(i)
        break
```

```python
#Q4: Find first repeating element
ls=['a','b','c','a','d','d']
m={}
for i in ls:
    m[i]=m.get(i,0)+1
    if m[i]>1:
        print(i)
        break
```

```python
#Q:repeating element
#Q: Input:
ls=["a","b","a","c","b","d"]
freq = {}
result = []
for i in ls:
    freq[i] = freq.get(i, 0) + 1
    if freq[i] == 2:
        result.append(i)

print(result)
```

```python
#Q: element appear only once
# Input:
ls = [1,2,3,2,1,4,5,4]
freq = {}
for i in ls:
    freq[i] = freq.get(i, 0) + 1

result = [i for i in ls if freq[i] == 1]

print(result)
```
